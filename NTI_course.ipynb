```python
prices=map(float,input("enter 3prices").split())
for prices in prices:
    if prices<=50:
        print(prices,"is cheap")
    elif prices>=100:
        print(prices,"is expensive")
    else:
        print(prices,"is mid")
```

```python
word=input("enter an even word")
result=word[::2]
mid=len(result)//2
result=result[:mid]+result[mid:][::-1]
result=result[-1]+result[1:-1]+result[0]
print(result.upper())
```

```python
grid = [[0] * 3 for _ in range(3)]
for i in range(3):
    for j in range(3):
     if i == j and i + j == 2:
            grid[i][j]
     elif i == j:
            grid[i][j] = 1
     elif i + j == 2:
            grid[i][j] = 2
for row in grid:
    print(row) 
```
